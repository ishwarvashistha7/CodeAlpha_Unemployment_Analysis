Get the dataset

```python
#Import libraries Used :-
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/content/Unemployment in India.csv')
df.head()
```

Understanding the dataset :-

```python
#Check the number of rows and columns :-
df.shape
```

```python
#Check all column names :-
df.columns
```

```python
#Check dataset information :-
df.info()
```

```python
#Statistical Summary :-
df.describe()
```

```python
#Check missing values :-
df.isnull().sum()
```

Data Cleaning :-

```python
#Check and clean columns :-
df.columns = df.columns.str.strip()  #Removed extra spaces from column names
df.columns
```

```python
#Clean extra spaces from text values :-
for col in df.select_dtypes(include='object').columns:
 df[col] = df[col].str.strip()
print("Text data cleaned successfully!")
```

```python
#check duplicate records :-
print("Duplicate rows :-",df.duplicated().sum())
```

```python
#Drop duplicate rows :-
df = df.drop_duplicates()
print("Duplicate rows dropped successfully!")
```

```python
#Handle missing values :-
df.replace(r'^\s*s',np.nan,regex=True,inplace=True)
df.isnull().sum()  #check missing values
```

```python
#Calculate percentage of missing values:-
missing_percentage = (df.isnull().sum()/len(df)*100)
missing_percentage
```

```python
#Check the cleaned dataset :-
df.head()
```

```python
#Find the missing row:-
missing_rows = df[df.isnull().any(axis=1)]
missing_rows
```

```python
#Check row completely empty:-
print("Completely empty row :-",df.isnull().sum())
```

```python
df = df.dropna(how='all')
print("missing values after removing empty rows",df.isnull().sum())
```

```python
df.isnull().all(axis=1).sum()
```

Data Cleaning completed

#Exploratory Data analysis :-

```python
#Convert Dataframe into datetime format :-
df['Date'] = pd.to_datetime(df['Date'],dayfirst = True)
```

```python
#Create Year and Month Columns
# Extract Year and Month
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month

# Display the first few rows
df[['Date', 'Year', 'Month']].head()
```

```python
#Calculate average unemployment rate :-
avg_unemployment = df['Estimated Unemployment Rate (%)'].mean()
print("Average unemployed rate :-", round(avg_unemployment, 2), "%")
```

```python
#Find the highest unemployment rate:-
highest = df.loc[df['Estimated Unemployment Rate (%)'].idxmax()]
highest
```

#DATA VISUALISTAION :-

```python
#Unemployment trend over time :-
#AVG Unemplyment rate each date
monthly_unemployment = df.groupby('Date')['Estimated Unemployment Rate (%)'].mean()
monthly_unemployment.head()
```

```python
#Graph Plot:-
plt.figure(figsize=(12, 6))
plt.plot(
    monthly_unemployment.index,
    monthly_unemployment.values,
    marker='o'
)
plt.title('Unemployment Rate Trend in India')
plt.xlabel('Date')
plt.ylabel('Unemployment Rate (%)')
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

```python
#StateWise unemplymment analysis :-
#States having highest unemployment
state_unemployment = df.groupby('Region')['Estimated Unemployment Rate (%)'].mean().sort_values(ascending=False)
state_unemployment
```

```python
#GRAPH PLOT :-
plt.figure(figsize=(12, 7))
sns.barplot(
    x=state_unemployment.values,
    y=state_unemployment.index
)
plt.title('Average Unemployment Rate by State')
plt.xlabel('Unemployment Rate (%)')
plt.ylabel('States')
plt.show()
```

```python
#COVID-19 IMPACT ANALYSIS :-
#For this we use March 2020 as starting point
before_covid = df[df['Date'] < '2020-03-01']
during_covid = df[df['Date'] >= '2020-03-01']
before_avg = before_covid['Estimated Unemployment Rate (%)'].mean()
during_avg = during_covid['Estimated Unemployment Rate (%)'].mean()
print("Before COVID:", round(before_avg, 2), "%")
print("During COVID:", round(during_avg, 2), "%")
```

```python
#Comparison Graph :-
plt.figure(figsize=(7, 5))
sns.barplot(
    x=['Before COVID', 'During COVID'],
    y=[before_avg, during_avg]
)
plt.title('Impact of COVID-19 on Unemployment')
plt.ylabel('Average Unemployment Rate (%)')
plt.show()
```

#RURAL V/s URBAN UNEMPLOYMENT :-

```python
#Compaing unemployment between rural and urban
area_unemployment = df.groupby('Area')['Estimated Unemployment Rate (%)'].mean()
area_unemployment
```

```python
#COMPARISON GRAPH PLOT :-
plt.figure(figsize=(8, 5))
sns.barplot(
    x=area_unemployment.index,
    y=area_unemployment.values
)
plt.title('Rural vs Urban Unemployment Rate')
plt.xlabel('Area')
plt.ylabel('Average Unemployment Rate (%)')
plt.show()
```

```python
#COVID-19 IMPACT ANALYSIS :-Unemmployment before covid-19 and during covid-19
before_covid = df[df['Date'] < '2020-03-01']
during_covid = df[df['Date'] >= '2020-03-01']
before_avg = before_covid['Estimated Unemployment Rate (%)'].mean()
during_avg = during_covid['Estimated Unemployment Rate (%)'].mean()
print("Before COVID:", round(before_avg, 2), "%")
print("During COVID:", round(during_avg, 2), "%")
```

```python
#COMPARISON GRAPH :-
plt.figure(figsize=(8, 5))
sns.barplot(
    x=['Before COVID', 'During COVID'],
    y=[before_avg, during_avg]
)
plt.title('COVID-19 Impact on Unemployment')
plt.ylabel('Average Unemployment Rate (%)')
plt.show()
```
